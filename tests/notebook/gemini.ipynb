```python
import os
import textwrap
from google.oauth2 import service_account
import vertexai
from vertexai.generative_models import GenerativeModel, GenerationConfig
from IPython.display import Markdown
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
GOOGLE_CLOUD_SERVICE_ACCOUNT_PATH=os.getenv("GOOGLE_CLOUD_SERVICE_ACCOUNT_PATH")
```

```python
credentials = service_account.Credentials.from_service_account_file(
    GOOGLE_CLOUD_SERVICE_ACCOUNT_PATH,
)
```

```python
print(credentials.service_account_email)
print(credentials.project_id)
```

```python
vertexai.init(project="silrokai", location="asia-northeast3", credentials=credentials)
```

```python
model = GenerativeModel(
    model_name="gemini-2.5-flash",
    generation_config=GenerationConfig(
        temperature=0.2,
        max_output_tokens=1024,
    )
)
```

```python
%%time
response = model.generate_content("삶이란 무엇이니? 짧게대답")
```

```python
def to_markdown(text):
    text = text.replace('•', '  *')
    return Markdown(textwrap.indent(text, '> ', predicate=lambda _: True))
```

```python
to_markdown(response.text)
```

```python
%%time
response = model.generate_content("삶이란 무엇이니? 짧게 대답", stream=True)
```

```python
for chunk in response:
    print(chunk.text)
    print("_"*80)
```
